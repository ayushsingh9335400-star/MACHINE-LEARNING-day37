# Logistic Regression 

```python
from sklearn.datasets import make_classification

import numpy as np 
X, y = make_classification(n_samples=100, n_features=2, n_informative=1, n_redundant=0,
                           n_classes=2, n_clusters_per_class=1, random_state=41,hypercube=False,class_sep=10)
```

```python
import matplotlib.pyplot as plt 
```

```python
plt.figure(figsize=(10,6))
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter', s=100)
```

```python
import numpy as np

def step(z):
    return 1 if z >= 0 else 0

def perceptron(X, y, lr=0.1, epochs=1000):
    # Add bias column of ones
    X = np.insert(X, 0, 1, axis=1)
    weights = np.ones(X.shape[1])

    for i in range(epochs):
        j = np.random.randint(0, len(y))  # pick a random sample
        y_hat = step(np.dot(X[j], weights))  # prediction
        error = y[j] - y_hat
        weights += lr * error * X[j]  # update rule

    return weights[0], weights[1:]  # bias, other weights
```

```python
intercept_,coef_ = perceptron(X,y)
```

```python
print(coef_)
print(intercept_)
```

```python
m = -(coef_[0]/coef_[1])
b = -(intercept_/coef_[1])
```

```python
x_input = np.linspace(-3,3,100)
y_input = m*x_input + b
```

```python
plt.figure(figsize=(10,6))
plt.plot(x_input,y_input,color='red',linewidth=3)
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=100)
plt.ylim(-3,2)
```

```python
from sklearn.linear_model import LogisticRegression
lor = LogisticRegression()
lor.fit(X,y)
```

```python
m = -(lor.coef_[0][0]/lor.coef_[0][1])
b = -(lor.intercept_/lor.coef_[0][1])
```

```python
x_input1 = np.linspace(-3,3,100)
y_input1 = m*x_input + b
```
